```python
import configparser
from email.mime.text import MIMEText
import logging
from pathlib import Path
import smtplib
from typing import Union
```

```python
dir_prj = Path.cwd().parent

# path to the config file
config_pth = dir_prj/'config'/'secrets.ini'

# load environment variables from the config file
config = configparser.ConfigParser()
config.read(config_pth)

# gmail credentials
send_user = config.get('DEFAULT', 'GMAIL_USER')
send_pass  = config.get('DEFAULT', 'GMAIL_PASSWORD')

send_user, send_pass
```

```python
def send_message(body: str, recepients: Union[str, list[str]]):
    """
    Send simple text messages to email, and phone numbers when using the correct address for the carrier.

    Args:
        body: Simple text to send.
        recepients: List of email addresses to send to.

    Verizon: ``3334445555@vtext.com``
    """
    # Create a MIMEText object with the body of the email.
    msg = MIMEText(body)
    
    # Set the sender's email.
    msg['From'] = send_user
    
    # Join the list of recipients into a single string separated by commas.
    msg['To'] = ', '.join(recipients)
       
    # Connect to Gmail's SMTP server using SSL.
    with smtplib.SMTP_SSL('smtp.gmail.com', 465) as smtp_server:
        
        # Login to the SMTP server using the sender's credentials.
        smtp_server.login(send_user, send_pass)
        
        # Send the email. The sendmail function requires the sender's email, the list of recipients, and the email message as a string.
        smtp_server.sendmail(send_user, recipients, msg.as_string())
        
    # Print a message to console after successfully sending the email.
    logging.debug(f"Message with text {body} sent to {msg['To']}.")

    return
```

```python
send_message('Message Content: https://www.recreation.gov/cart', "4173436390@vtext.com")
```
