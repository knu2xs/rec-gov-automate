```python
import pandas as pd

from rec_gov_automate.availability import get_campground_availability_by_month, filter_campsites_to_available_weekends
```

```python
# campground_id = 232464  # klaloch
# campground_id = 232094  # salmon la sac
# campground_id = 233877  # la wis wis
campground_id = 233950  # Pioneer Park
# campground_id = 233791  # Albeni Cove
start_year = 2025
start_month = 6
start_day = 1
```

```python
# get availability for May through September
campsite_df = pd.concat([get_campground_availability_by_month(campground_id, start_year, month) for month in range(5, 10)])

# filter to just sites available on the weekends, both Friday and Saturday night
campsite_df = filter_campsites_to_available_weekends(campsite_df)

# get available campsites that are not walk-in
campsite_df = campsite_df.loc[
    (campsite_df['campsite_type'] != 'WALK TO')
    & (~campsite_df['campsite_type'].str.contains('GROUP SHELTER'))
]

campsite_df
```
